1. Які категорії товарів приносять найбільший/найменший дохід у кожній філії?


```python
res1 = df.pivot_table(index='Product line', columns='Branch', values='gross income', aggfunc='sum')
res1
```

```python
def min_max_branch(val):
  return val.idxmin(), val.idxmax()
```

```python
res1[['min', 'max']] = res1.apply(
    min_max_branch,
    axis=1,
    result_type='expand'
)
```

```python
res1
```

```python
df.groupby('Branch').City.unique()
```

У місті Yangon (філія A) найбільший дохід приносять категорії Home and lifestyle та Sports and travel, а найменший — Health and beauty та Fashion accessories. У місті Mandalay (філія B) найбільший дохід приносять Health and beauty та Sports and travel, а найменший — Food and beverages та Electronic accessories. У місті Naypyitaw (філія C) найбільший дохід приносять Food and beverages, Fashion accessories та Electronic accessories, а найменший — Home and lifestyle та Sports and travel. На основі цих даних можна розглянути збільшення асортименту та запасів у категоріях з найвищим доходом і додатково дослідити причини низької дохідності категорій з найнижчими показниками.

2. Які категорії товарів мають найбільшу кількість проданих одиниць у різних філіях?

```python
res2 = df.pivot_table(index='Product line', columns='Branch', values='Quantity', aggfunc='sum')
res2
```

```python
res2[['min', 'max']] = res2.apply(
    min_max_branch,
    axis=1,
    result_type='expand'
)
```

```python
res2
```

Тепер ми бачимо цікаву(або очікуєму) картину, якщо порівняти дані результати із попередніми, де ми рахували які категорії товарів приносять дохід у кожній філії, вони майже збігаються. Тобто для всіх категорій філія з найменшим валовим доходом також є філією з найменшою кількістю проданих товарів. Якщо порівнювати найбільший дохід і найбільшу продану кільксть всі катергорії окрім 'Sports and travel' також співпадають.
Ще можна порівняти із зведеною таблицею що була в завданні 3.1 де ми дивились кількість інвойсів, тобто як часто купують товри певної категорії в кожній філії.

```python
invoices_by_category = invoices_by_category.T
invoices_by_category[['min', 'max']] = invoices_by_category.apply(
    min_max_branch,
    axis=1,
    result_type='expand'
)
```

```python
invoices_by_category
```

Порівняння кількості інвойсів, проданих одиниць товару та доходу показує загальну схожу тенденцію: категорії, які мають більшу кількість покупок та проданих товарів, переважно приносять і більший дохід. Водночас повного збігу між показниками немає, що може свідчити про вплив середньої вартості товарів у категорії.

```python
result = pd.DataFrame({
    'Invoices': invoices_by_category['min'] + ' / ' + invoices_by_category['max'],
    'Quantity': res2['min'] + ' / ' + res2['max'],
    'Income': res1['min'] + ' / ' + res1['max']
})
result
```

```python
result['Match'] = (
    (result['Invoices'] == result['Quantity']) &
    (result['Quantity'] == result['Income'])
)
```

```python
result
```

У чотирьох категоріях — Fashion Accessories, Food and Beverages, Health and Beauty та Home and Lifestyle — спостерігається однакова закономірність: одні й ті самі філії мають найвищі та найнижчі значення за кількістю продажів, кількістю проданих товарів і доходом.
