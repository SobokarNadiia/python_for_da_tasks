2. Напишіть функцію `calculation` яка приймає на вхід 4 аргументи типу int - `a`, `b`, `c`, `d` і вертає результат обчислення такої формули: $ (a + b - c) * d $.

  Викличте функцію на наступному наборі аргументів і виведіть результат на екран:
  ```
  a = 16
  b = 4
  c = 9

  ```

````python
def calculation(a, b, c, d):
  return round((a + b + c) / d, 2)

calculation(16, 4, 9, 3)
````

5. **Ділення з перевіркою типів.**  

  1. Напишіть функцію `int_division` яка приймає на вхід два аргументи типу `int` - `a` і `b`, і виконує ділення першого аргумента на другий: `a/b`. При цьому функція перед діленням перевіряє, чи другий аргумент не дорівнює 0. Якщо другий аргумент не 0, фукнція вертає результат ділення, інакше - замість виконання ділення, вертає повідомлення на екран "Не можу виконати ділення на 0".

    Якщо типи, надані у функцію, не `int` - функція виводить повідомлення "Не працюю з такими типами даних.".

  2. Запустіть функцію на наступних парах чисел:
- 12 і 6
- 20 і 0
- 112 і 10
- 99 і 0
- 11 i 1.1

    і виведіть на екран повідомлення "Результат роботи функції: " - і далі те, що повернула функція.

  3. Дайте відповідь на питання: "Чому в деяких випадках функція повернула None?"

````python
def int_division(a: int, b: int):
  if isinstance(a, int) and isinstance(b, int):
    if b == 0:
      print('Не можу виконати ділення на 0')
    else:
      return a / b
  else:
    print('Не працюю з такими типами даних.')
````

````python
numbers = [[12, 6], [20, 0], [112, 10], [99, 0], [11, 1.1]]

print('Результат роботи функції:')

for num in numbers:
 print(int_division(num[0], num[1]))
````

Чому в деяких випадках функція повернула None?

Тому що при деяких умовах ми не зазначали в функції, що вона має повертати і повертає за замовчуванням none.
